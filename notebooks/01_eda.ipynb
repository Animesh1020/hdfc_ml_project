```python
numeric_cols = [
    'Applicant_Income',
    'Loan_Amount',
    'Asset_Value',
    'Debt_to_Income_Ratio',
    'Annual_Household_Income'
]

for col in numeric_cols:
    print(col)
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print("Outliers:", outliers)
```
